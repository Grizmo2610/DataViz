```python
import pandas as pd
import random
from faker import Faker

fake = Faker("vi_VN")

n = random.randint(20, 50)

data = []

products = [
    "Laptop", "Điện thoại", "Tai nghe", "Bàn phím",
    "Màn hình", "Chuột", "Máy tính bảng"
]

payment_methods = ["Tiền mặt", "Chuyển khoản", "Thẻ tín dụng", "Ví điện tử"]
sale_statuses = ["Completed", "Pending", "Cancelled"]

for _ in range(n):
    quantity = random.randint(1, 5)
    unit_price = random.randint(500_000, 30_000_000)

    data.append({
        "sale_id": fake.uuid4(),
        "sale_date": fake.date_between(start_date="-1y", end_date="today"),
        "customer_name": fake.name(),
        "customer_email": fake.email(),
        "product_name": random.choice(products),
        "category": random.choice(["Electronics", "Accessories", "Computer"]),
        "quantity": quantity,
        "unit_price": unit_price,
        "discount_percent": random.choice([0, 5, 10, 15, 20]),
        "total_amount": quantity * unit_price,
        "salesperson": fake.name(),
        "region": random.choice(["North", "Central", "South"]),
        "payment_method": random.choice(payment_methods),
        "sale_status": random.choice(sale_statuses),
    })

df = pd.DataFrame(data)

df["total_amount"] = (
    df["quantity"]
    * df["unit_price"]
    * (1 - df["discount_percent"] / 100)
).round(0)

df.to_csv("sales_data.csv", index=False, encoding="utf-8-sig")

print(df)
print(f"\nĐã lưu {len(df)} dòng vào sales_data.csv")
```
